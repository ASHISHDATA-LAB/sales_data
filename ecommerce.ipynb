```python
#step import req libraries 
import pandas as pd 
import numpy as np
```

```python
#step import the dataset 
df = pd.read_excel("Ecommerce_Unclean_Project.xlsx")
df.head()
```

```python
#step: view the information
df.info()
```

```python
# statistical summary 
df.describe()
```

```python
# dataset shape
df.shape
```

```python
# step : check the missing values 
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# remove duplicate record
df.drop_duplicates(inplace=True)
```

```python
#replace invalid values 
#replace common invalid values with NaN 

df.replace(['N/A','NULL',' '],np.nan,inplace=True)
```

```python
# remove the blank
df = df.apply(lambda x: x.str.strip() if x.dtype == 'object' else x)
```

```python
# Standard Text Format
# Convert Text into Proper Case.
df['Category'] = df['Category'].str.title()
df['City'] = df['City'].str.title()
df['Payment_Mode'] = df['Payment_Mode'].str.title()
df['Product'] = df['Product'].str.title()
df['State'] = df['State'].str.title()
```

```python
# Validate Email Address
# keep only valid records

df = df[df['Email'].str.contains('@',na=False)]
```

```python
# convert data columnsabs
df['Order_Date'] = pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
# convert numeric columns

cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
df.head(20)
```

```python
# record invalid quantity 
# remoce record hoaving negative or zero 
df = df[df['Qty']>0]
```

```python
df.info()
```

```python
df['Discount'] = pd.to_numeric(df['Discount'],errors='coerce').fillna(0)
df['Phone'] = df['Phone'].fillna('Unknown')
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# feacture engineering 
# total sales 
# net sales
df['Total_sales'] = df['Unit_Price']*df['Qty']
df['Net_sales'] = df["Total_sales"]-(df['Total_sales']*df['Discount']/100)
```

```python
# 20% profit 
df['Profit'] = df['Net_sales']*20/100
```

```python
#order month
df["Order_month"] = df['Order_Date'].dt.month_name()
```

```python
#order Year
# weekday
df['Order_year'] = df['Order_Date'].dt.year
df['Weekday'] = df['Order_Date'].dt.day_name()
```

```python
df.info()
```

```python
df.describe()
```

```python
# Export clean data 
df.to_csv('Clean_Ecommerce.csv',index=False)
```
